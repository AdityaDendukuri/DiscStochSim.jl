```julia
using Catalyst, CommonSolve, JumpProcesses
using Plots, UnicodePlots, Distributions
using ExponentialUtilities, Expokit
include("src/DiscStochSim.jl")
using .DiscStochSim
```

```julia
rn = @reaction_network begin
	 @species A(t) B(t) C(t) 
	 @parameters k1 k2 k3
	 (k1, k2), A <--> B
	 k3      , B --> C + A  
end
```

```julia
#initial condition u0 
u0 = [:A => 100, :B => 0, :C => 0]

# time span
tspan = (0., 200.)

# rates
ps = [:k1 => 0.1, :k2 => 0.1, :k3 => 1e-3]
```

```julia
jprob = JumpProblem(rn, u0, tspan, ps)
jump_sol = solve(jprob)
plot(jump_sol; lw = 1)
```
