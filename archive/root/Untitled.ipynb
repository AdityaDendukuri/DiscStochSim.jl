```python
import numpy as np

n = 5
A = np.random.rand(n, n)
b = np.random.rand(n)

# solve Ax = b
x = np.linalg.solve(A, b)

# perturb b
b_new = b * (1 + 1e-4)

# solve again
x_new = np.linalg.solve(A, b_new)

# compare relative changes
rel_b = np.linalg.norm(b_new - b) / np.linalg.norm(b)
rel_x = np.linalg.norm(x_new - x) / np.linalg.norm(x)

print("rel change in b:", rel_b)
print("rel change in x:", rel_x)
print("cond(A):", np.linalg.cond(A))
```
